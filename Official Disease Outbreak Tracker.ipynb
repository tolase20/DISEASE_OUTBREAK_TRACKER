## Methodology

The program follows these steps:

1. Store the outbreak records in a list of dictionaries.
2. Validate each record before analysis.
3. Exclude invalid records from analytical totals.
4. Calculate confirmation, fatality, recovery, and active-case rates.
5. Classify each valid record as Low, Moderate, High, or Critical.
6. Allow users to search the records by country or disease.
7. Display High and Critical records for monitoring.
8. Generate a final summary and automated alerts.

## Data Validation Rules

A record is marked as invalid when:

- A numerical value is negative.
- Confirmed cases exceed suspected cases.
- Deaths exceed confirmed cases.
- Recovered cases exceed confirmed cases.
- Active cases exceed confirmed cases.

Invalid records remain in the data-quality report but are excluded from metrics, risk classification, summaries, and alerts.

## Key Metric Formulas

- Confirmation Rate = Confirmed Cases ÷ Suspected Cases × 100
- Case Fatality Rate = Deaths ÷ Confirmed Cases × 100
- Recovery Rate = Recovered Cases ÷ Confirmed Cases × 100
- Active Case Rate = Active Cases ÷ Confirmed Cases × 100

## Risk Classification Rules

- Critical: Confirmed cases are at least 500 and deaths are at least 20, or the case fatality rate is at least 10% with at least 100 confirmed cases.
- High: Confirmed cases are at least 200 or deaths are at least 10.
- Moderate: Confirmed cases are at least 50.
- Low: The record does not meet any of the higher conditions.

These thresholds are educational project rules. They are not official WHO or Africa CDC classifications.

## Search & Monitoring

* **`search_records()`** searches records by country or disease using an exact, case-insensitive match.
* Displays matching records with their **risk classification** and total count.
* Searches **both valid and invalid records**.
* **`show_high_priority()`** filters valid records classified as **HIGH or CRITICAL**.
* Reuses **`classify_risk()`** to ensure consistent risk assessment.
* Reports when no matches or high-priority records are found.


## Alert Engine Thresholds

The alert engine flags a valid record when one or more of the following conditions are met:

- The risk level is High or Critical.
- Deaths are at least 10.
- The confirmation rate is at least 80%.
- The active-case rate is at least 30%.

These are classroom thresholds selected for the prototype and can be adjusted if the project requirements change.

```python
"""Disease Outbreak Tracker

An educational decision-support prototype built with Python fundamentals.
All records are simulated and are not official epidemiological statistics.
"""

# Project thresholds. These are classroom rules, not WHO or Africa CDC standards

CRITICAL_CONFIRMED_THRESHOLD = 500
CRITICAL_DEATH_THRESHOLD = 20
CRITICAL_FATALITY_RATE = 10.0
CRITICAL_FATALITY_MIN_CASES = 100
HIGH_CONFIRMED_THRESHOLD = 200
HIGH_DEATH_THRESHOLD = 10
MODERATE_CONFIRMED_THRESHOLD = 50
HIGH_CONFIRMATION_RATE = 80.0
HIGH_ACTIVE_CASE_RATE = 30.0

# Each dictionary is one outbreak record. This list contains 23 simulated records

outbreaks = [
    {"outbreak_id": "OUT001", "country": "Nigeria", "state_region": "Lagos", "disease": "Cholera", "suspected_cases": 250, "confirmed_cases": 180, "deaths": 12, "recovered": 130, "active_cases": 38, "report_date": "2026-08-01"},
    {"outbreak_id": "OUT002", "country": "Ghana", "state_region": "Greater Accra", "disease": "Measles", "suspected_cases": 150, "confirmed_cases": 100, "deaths": 4, "recovered": 75, "active_cases": 21, "report_date": "2026-08-02"},
    {"outbreak_id": "OUT003", "country": "Kenya", "state_region": "Nairobi", "disease": "Cholera", "suspected_cases": 320, "confirmed_cases": 240, "deaths": 14, "recovered": 170, "active_cases": 56, "report_date": "2026-08-03"},
    {"outbreak_id": "OUT004", "country": "Uganda", "state_region": "Kampala", "disease": "Yellow Fever", "suspected_cases": 90, "confirmed_cases": 60, "deaths": 3, "recovered": 45, "active_cases": 12, "report_date": "2026-08-04"},
    {"outbreak_id": "OUT005", "country": "Nigeria", "state_region": "Kano", "disease": "Measles", "suspected_cases": 600, "confirmed_cases": 520, "deaths": 24, "recovered": 410, "active_cases": 86, "report_date": "2026-08-05"},
    {"outbreak_id": "OUT006", "country": "Sierra Leone", "state_region": "Western Area", "disease": "Cholera", "suspected_cases": 420, "confirmed_cases": 310, "deaths": 16, "recovered": 220, "active_cases": 74, "report_date": "2026-08-06"},
    {"outbreak_id": "OUT007", "country": "South Africa", "state_region": "Gauteng", "disease": "Influenza", "suspected_cases": 300, "confirmed_cases": 180, "deaths": 5, "recovered": 150, "active_cases": 25, "report_date": "2026-08-07"},
    {"outbreak_id": "OUT008", "country": "Liberia", "state_region": "Montserrado", "disease": "Lassa Fever", "suspected_cases": 130, "confirmed_cases": 105, "deaths": 13, "recovered": 70, "active_cases": 22, "report_date": "2026-08-08"},
    {"outbreak_id": "OUT009", "country": "Senegal", "state_region": "Dakar", "disease": "Measles", "suspected_cases": 75, "confirmed_cases": 50, "deaths": 2, "recovered": 38, "active_cases": 10, "report_date": "2026-08-09"},
    {"outbreak_id": "OUT010", "country": "Nigeria", "state_region": "Rivers", "disease": "Yellow Fever", "suspected_cases": 210, "confirmed_cases": 155, "deaths": 9, "recovered": 110, "active_cases": 36, "report_date": "2026-08-10"},
    {"outbreak_id": "OUT011", "country": "Cameroon", "state_region": "Centre", "disease": "Cholera", "suspected_cases": 700, "confirmed_cases": 580, "deaths": 27, "recovered": 420, "active_cases": 133, "report_date": "2026-08-11"},
    {"outbreak_id": "OUT012", "country": "Rwanda", "state_region": "Kigali", "disease": "Measles", "suspected_cases": 40, "confirmed_cases": 25, "deaths": 1, "recovered": 20, "active_cases": 4, "report_date": "2026-08-12"},
    {"outbreak_id": "OUT013", "country": "Ghana", "state_region": "Ashanti", "disease": "Cholera", "suspected_cases": 280, "confirmed_cases": 210, "deaths": 11, "recovered": 145, "active_cases": 54, "report_date": "2026-08-13"},
    {"outbreak_id": "OUT014", "country": "Zambia", "state_region": "Lusaka", "disease": "Cholera", "suspected_cases": 180, "confirmed_cases": 120, "deaths": 6, "recovered": 90, "active_cases": 24, "report_date": "2026-08-14"},
    {"outbreak_id": "OUT015", "country": "Ethiopia", "state_region": "Addis Ababa", "disease": "Measles", "suspected_cases": 500, "confirmed_cases": 350, "deaths": 18, "recovered": 260, "active_cases": 72, "report_date": "2026-08-15"},
   
    # These three records are intentionally invalid
    
    {"outbreak_id": "OUT016", "country": "Nigeria", "state_region": "Kaduna", "disease": "Cholera", "suspected_cases": 100, "confirmed_cases": 150, "deaths": 5, "recovered": 80, "active_cases": 65, "report_date": "2026-08-16"},
    {"outbreak_id": "OUT017", "country": "Ghana", "state_region": "Volta", "disease": "Measles", "suspected_cases": 100, "confirmed_cases": 70, "deaths": 80, "recovered": 40, "active_cases": 10, "report_date": "2026-08-17"},
    {"outbreak_id": "OUT018", "country": "Kenya", "state_region": "Mombasa", "disease": "Cholera", "suspected_cases": 200, "confirmed_cases": 150, "deaths": 5, "recovered": 100, "active_cases": 200, "report_date": "2026-08-18"},
    
    # Five student-created records. OUT023 is intentionally invalid
    
    {"outbreak_id": "OUT019", "country": "Botswana", "state_region": "Gaborone", "disease": "Influenza", "suspected_cases": 35, "confirmed_cases": 18, "deaths": 0, "recovered": 14, "active_cases": 4, "report_date": "2026-08-19"},
    {"outbreak_id": "OUT020", "country": "Tanzania", "state_region": "Dar es Salaam", "disease": "Cholera", "suspected_cases": 360, "confirmed_cases": 260, "deaths": 9, "recovered": 190, "active_cases": 61, "report_date": "2026-08-20"},
    {"outbreak_id": "OUT021", "country": "Nigeria", "state_region": "Oyo", "disease": "Measles", "suspected_cases": 230, "confirmed_cases": 170, "deaths": 7, "recovered": 120, "active_cases": 43, "report_date": "2026-08-21"},
    {"outbreak_id": "OUT022", "country": "Namibia", "state_region": "Khomas", "disease": "Yellow Fever", "suspected_cases": 140, "confirmed_cases": 95, "deaths": 3, "recovered": 70, "active_cases": 22, "report_date": "2026-08-22"},
    {"outbreak_id": "OUT023", "country": "Uganda", "state_region": "Gulu", "disease": "Lassa Fever", "suspected_cases": 80, "confirmed_cases": 110, "deaths": 6, "recovered": 50, "active_cases": 54, "report_date": "2026-08-23"},
]

# Finding problems with one record, such as negative or inconsistent case counts

def validation_errors(record):
    """Return a list of every project-rule violation in one record."""
    errors = []
    numeric_fields = ["suspected_cases", "confirmed_cases", "deaths", "recovered", "active_cases"]
    index = 0
    while index < len(numeric_fields):
        field = numeric_fields[index]
        if record[field] < 0:
            errors.append(field.replace("_", " ").title() + " cannot be negative.")
        index += 1
    if record["confirmed_cases"] > record["suspected_cases"]:
        errors.append("Confirmed cases exceed suspected cases.")
    if record["deaths"] > record["confirmed_cases"]:
        errors.append("Deaths exceed confirmed cases.")
    if record["recovered"] > record["confirmed_cases"]:
        errors.append("Recovered cases exceed confirmed cases.")
    if record["active_cases"] > record["confirmed_cases"]:
        errors.append("Active cases exceed confirmed cases.")
    return errors

# A record is valid if the problem list is empty

def is_valid(record):
    return len(validation_errors(record)) == 0

# Calculating the confirmation, fatality, recovery, and active-case percentages

def calculate_metrics(record):
    """Calculate required rates safely; return None for unavailable denominators."""
    suspected = record["suspected_cases"]
    confirmed = record["confirmed_cases"]
    confirmation_rate = None
    fatality_rate = None
    recovery_rate = None
    active_rate = None
    if suspected > 0:
        confirmation_rate = confirmed / suspected * 100
    if confirmed > 0:
        fatality_rate = record["deaths"] / confirmed * 100
        recovery_rate = record["recovered"] / confirmed * 100
        active_rate = record["active_cases"] / confirmed * 100
    return {
        "confirmation_rate": confirmation_rate,
        "case_fatality_rate": fatality_rate,
        "recovery_rate": recovery_rate,
        "active_case_rate": active_rate,
    }

# RISK CLASSIFICATION
# Applying the Low, Moderate, High, and Critical classroom rules

def classify_risk(record):
    """Return the analytical priority and transparent reason(s)."""
    metrics = calculate_metrics(record)
    fatality_rate = metrics["case_fatality_rate"]
    reasons = []
    critical_volume = record["confirmed_cases"] >= CRITICAL_CONFIRMED_THRESHOLD and record["deaths"] >= CRITICAL_DEATH_THRESHOLD
    critical_fatality = fatality_rate is not None and fatality_rate >= CRITICAL_FATALITY_RATE and record["confirmed_cases"] >= CRITICAL_FATALITY_MIN_CASES
    if critical_volume or critical_fatality:
        if critical_volume:
            reasons.append("Confirmed cases and deaths reached the critical volume thresholds.")
        if critical_fatality:
            reasons.append("Case fatality rate and confirmed cases reached the critical severity thresholds.")
        return "CRITICAL", reasons
    if record["confirmed_cases"] >= HIGH_CONFIRMED_THRESHOLD or record["deaths"] >= HIGH_DEATH_THRESHOLD:
        if record["confirmed_cases"] >= HIGH_CONFIRMED_THRESHOLD:
            reasons.append("Confirmed cases reached the high-case threshold.")
        if record["deaths"] >= HIGH_DEATH_THRESHOLD:
            reasons.append("Reported deaths reached the high-death threshold.")
        return "HIGH", reasons
    if record["confirmed_cases"] >= MODERATE_CONFIRMED_THRESHOLD:
        reasons.append("Confirmed cases reached the moderate-case threshold.")
        return "MODERATE", reasons
    reasons.append("The record did not reach the moderate, high, or critical thresholds.")
    return "LOW", reasons


def rate_text(value):
    if value is None:
        return "N/A (zero denominator)"
    return f"{value:.2f}%"


def display_record(record, show_risk=False):
    print("-" * 62)
    print(f'{record["outbreak_id"]} | {record["report_date"]}')
    print(f'{record["country"]} - {record["state_region"]} | {record["disease"]}')
    print(f'Suspected: {record["suspected_cases"]:,} | Confirmed: {record["confirmed_cases"]:,} | Deaths: {record["deaths"]:,}')
    print(f'Recovered: {record["recovered"]:,} | Active: {record["active_cases"]:,}')
    if show_risk:
        if is_valid(record):
            risk, reasons = classify_risk(record)
            print("Analytical Priority:", risk)
            index = 0
            while index < len(reasons):
                print("Reason:", reasons[index])
                index += 1
        else:
            print("Analytical Priority: NOT CALCULATED - INVALID RECORD")

# Printing one record and optionally explain its risk level

def view_all_records(records):
    print("\nALL OUTBREAK RECORDS (SIMULATED)")
    index = 0
    while index < len(records):
        display_record(records[index])
        index += 1
    print(f"\nTotal records displayed: {len(records)}")


def get_nonnegative_integer(prompt):
    while True:
        value = input(prompt).strip()
        if value.isdigit():
            return int(value)
        print("Please enter a whole number that is zero or greater.")

# Asking for a new record and add it only if it passes validation

def add_record(records):
    print("\nADD NEW OUTBREAK RECORD")
    outbreak_id = input("Enter outbreak ID: ").strip().upper()
    index = 0
    while index < len(records):
        if records[index]["outbreak_id"].upper() == outbreak_id:
            print("That outbreak ID already exists. Record was not added.")
            return
        index += 1
    if outbreak_id == "":
        print("Outbreak ID cannot be blank. Record was not added.")
        return
    new_record = {
        "outbreak_id": outbreak_id,
        "country": input("Enter country: ").strip().title(),
        "state_region": input("Enter state/region: ").strip().title(),
        "disease": input("Enter disease: ").strip().title(),
        "suspected_cases": get_nonnegative_integer("Enter suspected cases: "),
        "confirmed_cases": get_nonnegative_integer("Enter confirmed cases: "),
        "deaths": get_nonnegative_integer("Enter deaths: "),
        "recovered": get_nonnegative_integer("Enter recovered: "),
        "active_cases": get_nonnegative_integer("Enter active cases: "),
        "report_date": input("Enter report date (YYYY-MM-DD): ").strip(),
    }
    errors = validation_errors(new_record)
    if len(errors) > 0:
        print("\nDATA QUALITY ALERT - record was not added.")
        index = 0
        while index < len(errors):
            print("-", errors[index])
            index += 1
    else:
        records.append(new_record)
        print(f'{outbreak_id} was validated and added successfully.')

#SEARCH AND MONITORING
#search by country, search by disease,and high-risk filtering.
#SEARCH & HIGH-RISK MONITORING
def search_records(records, field, label):
    """Search all records (valid and invalid) by an exact, case-insensitive
    match on the given field (e.g. field="country", label="Country")."""
    term = input(f"Enter {label.lower()} to search for: ").strip()
    if term == "":
        print(f"\n{label} cannot be blank.")
        return

    print(f"\n{label.upper()} SEARCH RESULTS: {term.title()}")
    print("-" * 62)
    count = 0
    index = 0
    while index < len(records):
        record = records[index]
        if record[field].lower() == term.lower():
            display_record(record, show_risk=True)
            count += 1
        index += 1

    print(f"\nTotal matching records: {count}")
    if count == 0:
        print(f"No records found for that {label.lower()}.")


def show_high_priority(records):
    """Show every valid record classified HIGH or CRITICAL, using the same
    classify_risk() rules as the rest of the program."""
    print("\n=== HIGH-PRIORITY OUTBREAKS (CLOSE MONITORING) ===")
    found = False
    index = 0
    while index < len(records):
        record = records[index]
        if is_valid(record):
            risk, reasons = classify_risk(record)
            if risk in ("HIGH", "CRITICAL"):
                display_record(record, show_risk=True)
                found = True
        index += 1
    if not found:
        print("No HIGH or CRITICAL records found.")

# DATA VALIDATION
# Checking all records and explaining any invalid case counts

def validate_all(records):
    print("\n" + "=" * 62)
    print("DATA QUALITY REPORT")
    print("=" * 62)
    valid_count = 0
    invalid_count = 0
    index = 0
    while index < len(records):
        record = records[index]
        errors = validation_errors(record)
        if len(errors) == 0:
            print(f'{record["outbreak_id"]}: VALID')
            valid_count += 1
        else:
            print(f'\n{record["outbreak_id"]}: INVALID')
            error_index = 0
            while error_index < len(errors):
                print("  Problem:", errors[error_index])
                error_index += 1
            invalid_count += 1
        index += 1
    print("\n" + "-" * 62)
    print(f"Total Records: {len(records)}")
    print(f"Valid Records: {valid_count}")
    print(f"Invalid Records: {invalid_count}")

# Displaying four calculated rates for valid records only

def display_metrics(records):
    print("\nOUTBREAK METRICS - VALID RECORDS ONLY")
    index = 0
    while index < len(records):
        record = records[index]
        if is_valid(record):
            metrics = calculate_metrics(record)
            print("-" * 62)
            print(f'{record["outbreak_id"]} | {record["disease"]} | {record["country"]}')
            print("Confirmation Rate:", rate_text(metrics["confirmation_rate"]))
            print("Case Fatality Rate:", rate_text(metrics["case_fatality_rate"]))
            print("Recovery Rate:", rate_text(metrics["recovery_rate"]))
            print("Active Case Rate:", rate_text(metrics["active_case_rate"]))
        index += 1


def alert_reasons(record):
    risk, reasons = classify_risk(record)
    metrics = calculate_metrics(record)
    alerts = []
    if risk in ["HIGH", "CRITICAL"]:
        alerts.append("The analytical priority is " + risk + ".")
    if record["deaths"] >= HIGH_DEATH_THRESHOLD:
        alerts.append("Reported deaths reached the project alert threshold.")
    if metrics["confirmation_rate"] is not None and metrics["confirmation_rate"] >= HIGH_CONFIRMATION_RATE:
        alerts.append("The confirmation rate reached the project alert threshold.")
    if metrics["active_case_rate"] is not None and metrics["active_case_rate"] >= HIGH_ACTIVE_CASE_RATE:
        alerts.append("The active-case rate reached the project alert threshold.")
    return alerts, reasons

# HIGH-RISK OUTBREAK MONITORING
# Displaying valid records classified High or Critical

def view_high_risk(records):
    print("\n" + "=" * 62)
    print("HIGH-PRIORITY OUTBREAK MONITORING")
    print("=" * 62)
    count = 0
    index = 0
    while index < len(records):
        record = records[index]
        if is_valid(record):
            risk, reasons = classify_risk(record)
            if risk in ["HIGH", "CRITICAL"]:
                display_record(record, show_risk=True)
                count += 1
        index += 1
    print(f"\nHigh/Critical valid records: {count}")
    print("These are project priorities, not official public-health classifications.")

# Printing alerts for valid records that meet the project thresholds

def outbreak_alert_engine(records):
    print("\n" + "=" * 62)
    print("OUTBREAK ALERT ENGINE")
    print("=" * 62)
    count = 0
    index = 0
    while index < len(records):
        record = records[index]
        if is_valid(record):
            alerts, risk_reasons = alert_reasons(record)
            if len(alerts) > 0:
                risk, unused = classify_risk(record)
                print("\nOUTBREAK ALERT")
                print(f'{record["outbreak_id"]} | {record["disease"]} | {record["country"]} - {record["state_region"]}')
                print("Analytical Priority:", risk)
                alert_index = 0
                while alert_index < len(alerts):
                    print("-", alerts[alert_index])
                    alert_index += 1
                print("Recommended analytical action: Prioritize this record for surveillance-team review.")
                count += 1
        index += 1
    print(f"\nTotal alerts: {count}")
    print("Alerts support review only; they are not medical advice or official warnings.")


def valid_records(records):
    result = []
    index = 0
    while index < len(records):
        if is_valid(records[index]):
            result.append(records[index])
        index += 1
    return result


def increment_count(counts, key, amount=1):
    if key in counts:
        counts[key] += amount
    else:
        counts[key] = amount


def largest_key(counts):
    winner = "N/A"
    winner_value = -1
    keys = list(counts.keys())
    index = 0
    while index < len(keys):
        if counts[keys[index]] > winner_value:
            winner = keys[index]
            winner_value = counts[keys[index]]
        index += 1
    return winner, winner_value

# Adding cases and deaths, and count diseases, countries, and risk levels

def build_summary(records):
    valid = valid_records(records)
    totals = {"suspected": 0, "confirmed": 0, "deaths": 0, "recovered": 0, "active": 0}
    disease_records = {}
    disease_confirmed = {}
    disease_deaths = {}
    country_records = {}
    country_confirmed = {}
    risk_counts = {"LOW": 0, "MODERATE": 0, "HIGH": 0, "CRITICAL": 0}
    highest_cfr_id = "N/A"
    highest_cfr = -1.0
    index = 0
    while index < len(valid):
        record = valid[index]
        totals["suspected"] += record["suspected_cases"]
        totals["confirmed"] += record["confirmed_cases"]
        totals["deaths"] += record["deaths"]
        totals["recovered"] += record["recovered"]
        totals["active"] += record["active_cases"]
        increment_count(disease_records, record["disease"])
        increment_count(disease_confirmed, record["disease"], record["confirmed_cases"])
        increment_count(disease_deaths, record["disease"], record["deaths"])
        increment_count(country_records, record["country"])
        increment_count(country_confirmed, record["country"], record["confirmed_cases"])
        risk, reasons = classify_risk(record)
        risk_counts[risk] += 1
        cfr = calculate_metrics(record)["case_fatality_rate"]
        if cfr is not None and cfr > highest_cfr:
            highest_cfr = cfr
            highest_cfr_id = record["outbreak_id"]
        index += 1
    most_disease, most_disease_count = largest_key(disease_records)
    top_disease_cases, top_disease_cases_value = largest_key(disease_confirmed)
    top_disease_deaths, top_disease_deaths_value = largest_key(disease_deaths)
    top_country_cases, top_country_cases_value = largest_key(country_confirmed)
    top_country_records, top_country_records_value = largest_key(country_records)
    return {
        "total_records": len(records), "valid_records": len(valid), "invalid_records": len(records) - len(valid),
        "totals": totals, "risk_counts": risk_counts,
        "most_disease": most_disease, "most_disease_count": most_disease_count,
        "top_disease_cases": top_disease_cases, "top_disease_cases_value": top_disease_cases_value,
        "top_disease_deaths": top_disease_deaths, "top_disease_deaths_value": top_disease_deaths_value,
        "top_country_cases": top_country_cases, "top_country_cases_value": top_country_cases_value,
        "top_country_records": top_country_records, "top_country_records_value": top_country_records_value,
        "highest_cfr_id": highest_cfr_id, "highest_cfr": highest_cfr,
        "disease_records": disease_records, "country_confirmed": country_confirmed,
    }


# FINAL SUMMARY REPORT
# Showing valid totals, most reported disease, and the top country.

def generate_summary(records):
    summary = build_summary(records)
    totals = summary["totals"]
    risks = summary["risk_counts"]
    print("\n" + "=" * 62)
    print("DISEASE OUTBREAK SUMMARY - VALID RECORDS ONLY")
    print("=" * 62)
    print(f'Total Records: {summary["total_records"]}')
    print(f'Valid Records: {summary["valid_records"]}')
    print(f'Invalid Records Excluded from Totals: {summary["invalid_records"]}')
    print(f'Total Suspected Cases: {totals["suspected"]:,}')
    print(f'Total Confirmed Cases: {totals["confirmed"]:,}')
    print(f'Total Deaths: {totals["deaths"]:,}')
    print(f'Total Recovered: {totals["recovered"]:,}')
    print(f'Total Active Cases: {totals["active"]:,}')
    print(f'Low: {risks["LOW"]} | Moderate: {risks["MODERATE"]} | High: {risks["HIGH"]} | Critical: {risks["CRITICAL"]}')
    print(f'High/Critical Records: {risks["HIGH"] + risks["CRITICAL"]}')
    print(f'Most Reported Disease: {summary["most_disease"]} ({summary["most_disease_count"]} records)')
    print(f'Disease With Highest Confirmed Cases: {summary["top_disease_cases"]} ({summary["top_disease_cases_value"]:,})')
    print(f'Disease With Highest Deaths: {summary["top_disease_deaths"]} ({summary["top_disease_deaths_value"]:,})')
    print(f'Country With Highest Confirmed Cases: {summary["top_country_cases"]} ({summary["top_country_cases_value"]:,})')
    print(f'Country With Most Records: {summary["top_country_records"]} ({summary["top_country_records_value"]})')
    print(f'Highest Case Fatality Rate: {summary["highest_cfr_id"]} ({summary["highest_cfr"]:.2f}%)')
    print("\nResponsible interpretation: Results describe simulated training data only.")

# ADVANCED PORTFOLIO EXTENSION: DATA QUALITY SCORE
# Each record starts at 100 points
# The program subtracts 20 points for each validation problem

def data_quality_score(record):
    """Advanced extension: 100 minus 20 points per validation-rule violation."""
    score = 100 - len(validation_errors(record)) * 20
    if score < 0:
        score = 0
    return score

# Printing the data-quality score for every record

def show_data_quality_scores(records):
    print("\nDATA QUALITY SCORE EXTENSION")
    print("Score = 100 minus 20 points for each failed validation rule.")
    index = 0
    while index < len(records):
        record = records[index]
        print(f'{record["outbreak_id"]}: {data_quality_score(record)}/100')
        index += 1

# =========================================================
# BONUS CHALLENGE
# This section tests whether the program updates when:
# 1. A record is removed
# 2. Confirmed cases are changed
# 3. Deaths are changed
# 4. A country is changed
# The test uses a copied dataset, so the original records
# are not permanently changed.
# =========================================================

def run_bonus_challenge(records):
    """Demonstrate the four bonus tests without changing the original dataset."""
    test_records = []
    index = 0
    while index < len(records):
        test_records.append(records[index].copy())
        index += 1

    print("\n" + "=" * 62)
    print("BONUS CHALLENGE TEST")
    print("=" * 62)
    print("A copy of the dataset is used, so the original records stay unchanged.")

    # Test 1: remove a record and check whether the summary changes.
    before_summary = build_summary(test_records)
    removed_record = test_records.pop(0)
    after_summary = build_summary(test_records)
    print("\n1. REMOVE A RECORD")
    print("Removed:", removed_record["outbreak_id"])
    print("Records before removal:", before_summary["total_records"])
    print("Records after removal:", after_summary["total_records"])
    print("Confirmed cases before removal:", before_summary["totals"]["confirmed"])
    print("Confirmed cases after removal:", after_summary["totals"]["confirmed"])

    # Test 2: change confirmed cases and check whether risk changes.
    risk_record = None
    index = 0
    while index < len(test_records):
        if test_records[index]["outbreak_id"] == "OUT007":
            risk_record = test_records[index]
            break
        index += 1

    old_risk, old_reason = classify_risk(risk_record)
    old_confirmed = risk_record["confirmed_cases"]
    risk_record["confirmed_cases"] = 220
    new_risk, new_reason = classify_risk(risk_record)
    print("\n2. CHANGE CONFIRMED CASES")
    print("Outbreak:", risk_record["outbreak_id"])
    print("Confirmed cases changed from", old_confirmed, "to", risk_record["confirmed_cases"])
    print("Risk changed from", old_risk, "to", new_risk)

    # Test 3: change deaths and check whether the fatality rate changes.
    fatality_record = None
    index = 0
    while index < len(test_records):
        if test_records[index]["outbreak_id"] == "OUT002":
            fatality_record = test_records[index]
            break
        index += 1

    old_deaths = fatality_record["deaths"]
    old_fatality_rate = calculate_metrics(fatality_record)["case_fatality_rate"]
    fatality_record["deaths"] = 10
    new_fatality_rate = calculate_metrics(fatality_record)["case_fatality_rate"]
    print("\n3. CHANGE DEATHS")
    print("Outbreak:", fatality_record["outbreak_id"])
    print("Deaths changed from", old_deaths, "to", fatality_record["deaths"])
    print("Case fatality rate changed from", rate_text(old_fatality_rate),
          "to", rate_text(new_fatality_rate))

    # Test 4: change a country and check whether the search result changes.
    country_record = None
    index = 0
    while index < len(test_records):
        if test_records[index]["outbreak_id"] == "OUT014":
            country_record = test_records[index]
            break
        index += 1

    old_country = country_record["country"]
    country_record["country"] = "Nigeria"
    nigeria_matches = 0
    index = 0
    while index < len(test_records):
        if test_records[index]["country"].lower() == "nigeria":
            nigeria_matches += 1
        index += 1
    print("\n4. CHANGE COUNTRY")
    print("Outbreak:", country_record["outbreak_id"])
    print("Country changed from", old_country, "to", country_record["country"])
    print("The updated country search finds", nigeria_matches, "Nigeria records.")

    print("\nBONUS RESULT: PASSED")
    print("The summary, risk classification, fatality rate, and search results")
    print("all update automatically when the copied data changes.")
    print("The original 23-record dataset was not changed.")


def analysis_menu(records):
    """Nested while-loop challenge."""
    while True:
        print("\nANALYSIS MENU")
        print("1. Disease and location summary")
        print("2. Outbreak alert engine")
        print("3. Data-quality score extension")
        print("4. Generate full summary")
        print("5. Run bonus challenge test")
        print("6. Return to main menu")
        choice = input("Enter your choice: ").strip()
        if choice == "1":
            summary = build_summary(records)
            print(f'\nMost reported disease: {summary["most_disease"]} ({summary["most_disease_count"]})')
            print(f'Highest-volume disease: {summary["top_disease_cases"]} ({summary["top_disease_cases_value"]:,} confirmed)')
            print(f'Highest-death disease: {summary["top_disease_deaths"]} ({summary["top_disease_deaths_value"]:,} deaths)')
            print(f'Highest-volume country: {summary["top_country_cases"]} ({summary["top_country_cases_value"]:,} confirmed)')
            print(f'Most represented country: {summary["top_country_records"]} ({summary["top_country_records_value"]} records)')
        elif choice == "2":
            outbreak_alert_engine(records)
        elif choice == "3":
            show_data_quality_scores(records)
        elif choice == "4":
            generate_summary(records)
        elif choice == "5":
            run_bonus_challenge(records)
        elif choice == "6":
            break
        else:
            print("Invalid choice. Enter a number from 1 to 6.")

# MAIN MENU
# Presenting options 1 through 9 and accepts the user choice

def main():
    while True:
        print("\n" + "=" * 62)
        print("DISEASE OUTBREAK TRACKER")
        print("Educational prototype using simulated data")
        print("=" * 62)
        print("1. View all outbreak records")
        print("2. Add new outbreak record")
        print("3. Search by country")
        print("4. Search by disease")
        print("5. Validate outbreak data")
        print("6. Calculate outbreak metrics")
        print("7. View high-risk outbreaks")
        print("8. Analysis menu and summary")
        print("9. Exit")
        choice = input("Enter your choice: ").strip()
        if choice == "1":
            view_all_records(outbreaks)
        elif choice == "2":
            add_record(outbreaks)
        elif choice == "3":
            search_records(outbreaks, "country", "Country")
        elif choice == "4":
            search_records(outbreaks, "disease", "Disease")
        elif choice == "5":
            validate_all(outbreaks)
        elif choice == "6":
            display_metrics(outbreaks)
        elif choice == "7":
            view_high_risk(outbreaks)
        elif choice == "8":
            analysis_menu(outbreaks)
        elif choice == "9":
            print("Thank you for using the Disease Outbreak Tracker.")
            break
        else:
            print("Invalid choice. Enter a number from 1 to 9.")


# PROGRAM SUCCESSFULLY lAUNCHED
# Running this file starts the program and displays the main menu
if __name__ == "__main__":
    main()
```
